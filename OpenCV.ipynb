## Basic Operations

### *. Load Image, Show Image, Convert the color/channel of image and Save the image
1. Read an image from file using cv2.imread().
2. Image read in BGR format.
3. We use cv2.imshow() for image showing.
4. If we want to change the channel of the image we call this function according to the color or channel, let we convert bgr to grayscale then, gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY).
5. After performing the changes any image we want to save these changes, so we called imwrite() funcion in this function we give arguments like output image file name as string and updated image file name as object where we do perform changes.

```python
import numpy as np
import cv2
```

```python
img = cv2.imread('i1.png')

if img is not None:
    print("Image Loaded Succesfully")
    cv2.imshow('Image',img) # It show the loaded image
    cv2.waitKey(0) # By this we can hold the image window, image window opens untill we don't press any key.
    cv2.destroyAllWindows() # It is use for to clean our environment and memory after completing it's work 
else:
    print("Error")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
if gray is not None:
    print("BGR Image is converted to grayscale Successfully!")
    cv2.imshow('GrayScale of Image',gray) 
    cv2.waitKey(0) 
    cv2.destroyAllWindows() 
else:
    print("Error Occured!")
```

```python
cv2.imwrite("Output.png",gray)
```

### Basic mini task:-
1. WAP to collect the file name or path from the user.
2. Load the image from the path.
3. Convert the bgr to grayscale.
4. Show the image.
5. Save the file as input user's given name for the file.
6. Finished


```python
import cv2

class imgRead():
    # Take input file name or filepath as string from the user
    def getFile(self,f_name):
        
        img = cv2.imread(f_name)
        print("Image Loaded Successfully!")
        return img
    
    # Convert the image color channerl bgr to graysclae
    def gray_cnvrt(self,img):
        gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
        return gray
    
    # Display the converted image
    def displayFile(self,img):
        if img is not None:                
            print("Image showing successfully!")
            cv2.imshow('Image', img)
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        else:
            print("Sorry!, Image Showing is failed.")
    
    # Here saved the modified file with custom name which is given by user for file
    def savefile(self,gray):
        name = input("Enter the custom file for saving this edited image:")
        cv2.imwrite(name,gray)
        return "File Saved Successfully!"
    

im = imgRead()

# Take file name or path
f = input("Enter file path:")
img = im.getFile(f)

# Convert bgr to gray
gray = im.gray_cnvrt(img)

# Display the image
im.displayFile(gray)

# Save the file 
im.savefile(gray)
    

```
