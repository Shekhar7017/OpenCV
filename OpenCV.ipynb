### Basic mini task:-
1. WAP to collect the file name or path from the user.
2. Load the image from the path.
3. Convert the bgr to grayscale.
4. Show the image.
5. Save the file as input user's given name for the file.
6. Finished


```python
import cv2

class imgRead():
    # Take input file name or filepath as string from the user
    def getFile(self,f_name):
        
        img = cv2.imread(f_name)
        print("Image Loaded Successfully!")
        return img
    
    # Convert the image color channerl bgr to graysclae
    def gray_cnvrt(self,img):
        gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
        return gray
    
    # Display the converted image
    def displayFile(self,img):
        if img is not None:                
            print("Image showing successfully!")
            cv2.imshow('Image', img)
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        else:
            print("Sorry!, Image Showing is failed.")
    
    # Here saved the modified file with custom name which is given by user for file
    def savefile(self,gray):
        name = input("Enter the custom file for saving this edited image:")
        cv2.imwrite(name,gray)
        return "File Saved Successfully!"
    

im = imgRead()

# Take file name or path
f = input("Enter file path:")
img = im.getFile(f)

# Convert bgr to gray
gray = im.gray_cnvrt(img)

# Display the image
im.displayFile(gray)

# Save the file 
im.savefile(gray)
    

```

## Image Resize, Rotation & Flipping 


### 1. Resize the image

```python
img = cv2.imread('i1.png')

h,w = img.shape[:2]

print('Original Image Pixel:',h,w)

resize = cv2.resize(img,(400,600))

h1,w1 = resize.shape[:2]
cv2.imshow('Original Image',img)
cv2.imshow('Rsized Image',resize)


cv2.waitKey(0)
cv2.destroyAllWindows()

print("Image resized Successfully....")
print("Resized Image Pixel:",h1,w1)

```

### Rotate the image


```python
c = (w//2,h//2)

m = cv2.getRotationMatrix2D(c,90,1.0)
rotated = cv2.warpAffine(img,m,(w,h))

cv2.imshow('Rotated Image',rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Flipping the image

```python
import cv2

img = cv2.imread('i1.png')

if img is not None:
    f1 = cv2.flip(img,0) # Vertical Flip
    f2 = cv2.flip(img,1) # Horizontal Flip
    f3 = cv2.flip(img,-1) # 180deg Flip
    
    cv2.imshow('Original image', img)
    cv2.imshow('Horizontal Flip', f2)
    cv2.imshow('Vertical Flip', f1)
    cv2.imshow('180deg Flip', f3)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Sorry! Image does not flipped properly")
    
    
```

### Cropping the image


```python
# First initialize the pixels in x,y
x1,y1 = 100,80
x2,y2 = 400,300

cropped = img[y1:y2,x1:x2]

cv2.imshow('original image',img)
cv2.imshow('cropped image',cropped)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Basic Mini Task 2:-
1. Take Inpute file from user.
2. Access the file.
3. Show Option for user (Draw Line, Reactangle, Circle, Crop Image)
4. Take coordinates and color.
5. Display Output and ask for saving this output.

```python
import cv2

# Define Function without parameter for taking coordinates
def takingTerms():
    file = input("Enter File Name or Path:")
    global img,co1, co2, color, center, shape
    img = cv2.imread(file)
    

    if img is not None:
        print("Image Is loaded Succesfully!")
    else:
        print("Oops!,Image not loaded during some technique errors")
    
    shape = input("Enter the name of shape in which of them (Line, Rectangle, Circle):").lower()
    
    if shape == 'line' or shape =='rectangle':
        print("Enter Coordinates (x1,y1)")
        x1 = int(input("Enter Values of x1:"))
        y1 = int(input("Enter Values of y1:"))
            
        print("Enter Coordinates (x2,y2)")
        x2 = int(input("Enter Values of x2:"))
        y2 = int(input("Enter Values of y2:"))
        co1 = (x1,y1)
        co2 = (x2,y2)
    
    else:
        print("Enter Position of Center's coordinates of Circle.")
        x = int(input("Enter Values of x:"))
        y = int(input("Enter Values of y:"))
        center = (x,y)
        
    
    
    
    print("Enter the Color of Shape in bgr format of BLUE,RED,GREEN (BGR)")
    b = int(input("Enter the quantity of Blue:"))
    g = int(input("Enter the quantity of Green:"))
    r = int(input("Enter the quantity of Red:"))

    color = (b,g,r)
    return img, co1, co2, color
    


# Define the function with parameters to draw a shape on image and display the image.
def drawShape(shape,color,img,co1,co2):
    if shape == 'line':
        print("Line Drawing!")
        line = cv2.line(img,co1,co2,color,2)
        cv2.imshow("Line",line)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    elif shape == 'rectangle!':
        print("Rectangle Drawing!")
        rec = cv2.rectangle(img,co1,co2,color,2)
        cv2.imshow("Rectangle",rec)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    elif shape == 'circle':
        r = int(input("Enter the radius of circle:"))
        
        cir = cv2.circle(img,center,r,color,2)
        print('Circle Drawing!')
        cv2.imshow("Circle",cir)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
        
    else:
        print("Enter Correct shape name please")
        
        
print("Welcome to the Drawing the shape on the image!")
takingTerms()

drawShape(shape,color,img,co1,co2)
print("Thank you for using this!")

    
```

# Video processing

```python
import cv2

cap = cv2.VideoCapture(0) # It access the default webcam

while True:
    ret, frame = cap.read()
    
    if not ret:
        print("Could not load Frame!")
        break
    
    cv2.imshow('Webcam Feed', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Outing......")
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

camera = cv2.VideoCapture(0) # It access the default webcam

w = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

forcc = cv2.VideoWriter_fourcc(*'mp4v')

recorder = cv2.VideoWriter('my_vid.mp4',forcc,20,(w,h))
while True:
    ret, frame = camera.read()
    
    if not ret:
        print("Could not load Frame!")
        break
    
    recorder.write(frame)
    cv2.imshow('Webcam Feed', frame)
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Outing......")
        break

camera.release()
recorder.release()
cv2.destroyAllWindows()
```
